# Atomic Data Summary Table

Build and export the summary table from a Carsus HDF file. Each displayed ion stage must have both genuine level data and line data. Level and line totals include only those displayed stages; gaps are shown as separate comma-delimited runs.

```python
from pathlib import Path
import sys

CARSUS_ROOT = Path("..").resolve()
if str(CARSUS_ROOT) not in sys.path:
    sys.path.insert(0, str(CARSUS_ROOT))

import carsus
from export_table import exporttable

carsus.exporttable = exporttable
```

```python
hdf_path = CARSUS_ROOT / "docs/kurucz_cd23_chianti_He_cmfgen_H_Si_I-II.h5"
summary = carsus.exporttable(hdf_path, journal="aas")
carsus.exporttable(hdf_path, journal="aa")
carsus.exporttable(hdf_path, journal="mnras")
carsus.exporttable(hdf_path, journal="nature")
summary
```

```python
reference_hdf_path = CARSUS_ROOT / "docs/kurucz_cd23_cmfgen_H_Si.h5"
reference_summary = carsus.exporttable(
    reference_hdf_path, journal="aas", output_stem="tardis_table"
)
reference_summary
```
