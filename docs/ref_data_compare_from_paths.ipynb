```python
import os
import shutil
import tempfile
from pathlib import Path

from tardisbase.testing.regression_comparison.analyzers import HDFComparator
```

```python
ref1 = Path(os.environ['REF1_PATH']).resolve()
ref2 = Path(os.environ['REF2_PATH']).resolve()

tmp = Path(tempfile.mkdtemp())
dir1 = tmp / 'ref1'
dir2 = tmp / 'ref2'
dir1.mkdir()
dir2.mkdir()

shared_name = 'unit_test_data.h5'
shutil.copy(ref1, dir1 / shared_name)
shutil.copy(ref2, dir2 / shared_name)
```

```python
result = HDFComparator(print_path=True).summarise_changes_hdf(shared_name, str(dir1), str(dir2))
result
```

```python
status = 'REFDATA COMPARISON FAILED' if result else 'REFDATA COMPARISON PASSED'
(ref1.parent.parent / 'refdata_compare_result').write_text(status)
print(status)
```
